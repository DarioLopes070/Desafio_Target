2. Faça um programa onde eu possa lançar movimentações de estoque dos produtos que estão no json abaixo, dando entrada ou saída da mercadoria no meu depósito, onde cada movimentação deve ter:
    • Um número identificador único.
    • Uma descrição para identificar o tipo da movimentação realizada
E que ao final da movimentação me retorne a qtde final do estoque do produto movimentado.

```python
dados_estoque = {
	"estoque":
	[
	  {
		"codigoProduto": 101,
		"descricaoProduto": "Caneta Azul",
		"estoque": 150
	  },
	  {
		"codigoProduto": 102,
		"descricaoProduto": "Caderno Universitário",
		"estoque": 75
	  },
	  {
		"codigoProduto": 103,
		"descricaoProduto": "Borracha Branca",
		"estoque": 200
	  },
	  {
		"codigoProduto": 104,
		"descricaoProduto": "Lápis Preto HB",
		"estoque": 320
	  },
	  {
		"codigoProduto": 105,
		"descricaoProduto": "Marcador de Texto Amarelo",
		"estoque": 90
	  }
	]
}
class controleEstoque:
    def __init__(self, dados_estoque):
        self.estoque = dados_estoque["estoque"]
        self.movimetacoes = []
       
    def movimentar(self, codigo_produto, tipo, quantidade, descricao):
        produto = next((item for item in self.estoque if item["codigoProduto"] == codigo_produto), None)
        
        if not produto:
            return {"erro": f"Produto com código {codigo_produto} não encontrado."}
        if tipo == "saida" and produto["estoque"] < quantidade:
            return {"erro": f"Estoque insuficiente para o produto {produto['descricaoProduto']}."}
        
        if tipo == "saida":
            produto["estoque"] -= quantidade
        elif tipo == "entrada":
            produto["estoque"] += quantidade
            
        nova_movimentacao = {
            "id": len(self.movimetacoes) + 1,
            "codigoProduto": codigo_produto,
            "descricao": descricao,
            "tipo": tipo,
            "quantidadeMovimentada": quantidade,
            "estoqueAtual": produto["estoque"]
        }    
        
        self.movimetacoes.append(nova_movimentacao)       
        return nova_movimentacao
              
def exibir_codigos_produtos(estoque):
    print("\n--- CÓDIGOS DE PRODUTOS ---")
    for produto in estoque:
        print(f"Código: {produto['codigoProduto']}, Descrição: {produto['descricaoProduto']}")              
sistema_estoque = controleEstoque(dados_estoque)
while True:
    print("\n--- MENU DE MOVIMENTAÇÃO ---")
    print("1. Registrar Entrada")
    print("2. Registrar Saída")
    print("3. Sair")
    opcao = input("Escolha uma opção: ")
    if opcao == "1":
        exibir_codigos_produtos(dados_estoque["estoque"])
        codigo = int(input("Digite o código do produto: "))
        quantidade = int(input("Digite a quantidade de entrada: "))
        descricao = input("Digite a descrição da movimentação: ")
        resultado = sistema_estoque.movimentar(codigo, "entrada", quantidade, descricao)
        print(f"\n{resultado}")
    elif opcao == "2":
        exibir_codigos_produtos(dados_estoque["estoque"])
        codigo = int(input("Digite o código do produto: "))
        quantidade = int(input("Digite a quantidade de saída: "))
        descricao = input("Digite a descrição da movimentação: ")
        resultado = sistema_estoque.movimentar(codigo, "saida", quantidade, descricao)
        print(f"\n{resultado}")
    elif opcao == "3":
        break
    else:
        print("Opção inválida. Tente novamente.")
```

3. Faça um programa que a partir de um valor e de uma data de vencimento, calcule o valor dos juros na data de hoje considerando que a multa seja de 2,5% ao dia.

```python
from datetime import datetime
def calcular_juros(valor_inicial, data_vencimento):
    data_vencimento = datetime.strptime(data_vencimento, "%Y-%m-%d").date()
    hoje = datetime.now().date()
    
    dias_atraso = (hoje - data_vencimento).days
    
    if dias_atraso <= 0:
        return {"valor_total": valor_inicial, "valor_juros": 0.0, "dias_atraso": 0}
    
    taxa_juros_diario = 0.025
    valor_juros = valor_inicial * taxa_juros_diario * dias_atraso
    valor_total = valor_inicial + valor_juros
    
    return {"valor_total": round(valor_total, 2), "valor_juros": round(valor_juros, 2), "dias_atraso": dias_atraso}


print("\n--- CÁLCULO DE JUROS ---")
valor_inicial = float(input("Digite o valor inicial da dívida: R$ "))
data_vencimento = input("Digite a data de vencimento (YYYY-MM-DD): ")
resultado = calcular_juros(valor_inicial, data_vencimento)
print(f"\n--- RESULTADO ---")
print(f"Valor total: R$ {resultado['valor_total']}")
print(f"Valor dos juros: R$ {resultado['valor_juros']}")
print(f"Dias de atraso: {resultado['dias_atraso']}")
```
